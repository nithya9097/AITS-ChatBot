```python
def HollowSquare(n):
    for i in range(n):
        for j in range(n):
            print("*" if i==0 or i==n-1 or j==0 or j==n-1 else " ", end="")
        print()
HollowSquare(5)
```

```python
n = 29

if n < 2:
    print("Not Prime")
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            print("Not Prime")
            break
    else:
        print("Prime")
```
